```python
from collections import deque
```

```python
maze = [
    ['S',' ','#',' ',' '],
    ['#',' ','#',' ','#'],
    [' ',' ',' ',' ',' '],
    [' ','#','#',' ',' '],
    [' ',' ',' ','#','G']
]
```

```python
row=len(maze)
```

```python
col = len(maze[0])
```

```python
start = (0,0)
```

```python
goal = (4,4)
```

```python
dir=[
    (-1,0),(1,0),(0,-1),(0,1)
]
```

```python
def bfs():
  queue = deque()
  queue.append((start,[start]))
  visited = set()
  visited.add(start)
  while queue:
    (x,y),path=queue.popleft()
    if(x,y)==goal:
      return path
    for dx,dy in dir:
      nx = x+dx
      ny = y+dy
      if(0<=nx<row and 0<=ny<col and maze[nx][ny]!='#' and (nx,ny) not in visited):
        visited.add((nx,ny))
        queue.append(((nx,ny),path+[(nx,ny)]))
  return None
```

```python
path=bfs()
```
